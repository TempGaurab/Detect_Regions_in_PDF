```python
from ultralytics import YOLO

# 1. Load a pretrained model (YOLOv8 Nano is great for small objects like checkboxes)
model = YOLO("yolov8n.pt") 
model
```

```python
# 2. Train the model
# 'data' points to your .yaml config file
results = model.train(
    data="path/to/your_data.yaml", 
    epochs=100, 
    imgsz=640, 
    batch=16,
    name="document_element_detector"
)
```

```python
from ultralytics import YOLO

# Load your best trained model
model = YOLO("runs/detect/document_element_detector/weights/best.pt")

# Run detection on an image
results = model.predict(source="test_document.png", save=True, conf=0.25)

# Extract coordinates for your evaluation script
for result in results:
    boxes = result.boxes.xyxy.cpu().numpy()  # [x1, y1, x2, y2]
    classes = result.boxes.cls.cpu().numpy() # Class IDs
    confidences = result.boxes.conf.cpu().numpy()
    
    print(f"Detected {len(boxes)} elements.")
```
