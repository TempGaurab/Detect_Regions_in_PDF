```python

"""
Hybrid Detection Framework — YOLOv11 + Advanced OpenCV
-------------------------------------------------------
YOLO acts as the primary backbone.
OpenCV acts as a post-processing validation layer:
  • High-confidence YOLO predictions  → accepted directly.
  • Low-confidence YOLO predictions   → cross-referenced with OpenCV candidates;
      confirmed overlaps are accepted, unresolved ones pass geometric filtering.
  • OpenCV-only detections            → supplement YOLO where nothing was found.
"""

import os
import cv2
import json
import math
import numpy as np
from PIL import Image
from ultralytics import YOLO

# ─────────────────────────────────────────────
# CONFIG
# ─────────────────────────────────────────────
WEIGHTS_PATH    = "runs/detect/run_v26l/runs/train/form_detector/weights/best.pt"   # ← YOLOv11 weights
IMAGE_FOLDER    = "data"
GT_FOLDER       = "clicking_mechanism/annotations"        # for evaluation
OUTPUT_FOLDER   = "hybrid_output"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Detection thresholds
CONF_HIGH       = 0.50    # above this  → accept YOLO directly
CONF_LOW        = 0.25    # below this  → discard YOLO even after opencv check
                          # between low and high → cross-reference with OpenCV

# Overlap threshold (IoU) to confirm a low-conf YOLO box via OpenCV
IOU_CONFIRM     = 0.30

# Tolerances for final evaluation
TOLERANCES      = [0.05, 0.10, 0.20]
CLASS_NAMES     = {0: "Checkboxes", 1: "Lines", 2: "Boxes"}

# ── Advanced OpenCV parameters ────────────────
CHECKBOX_MIN      = 10
CHECKBOX_MAX      = 50
SQUARE_TOLERANCE  = 0.30
LINE_MIN_W        = 100
LINE_MAX_H        = 4
LABEL_SEARCH_PX   = 300
LABEL_PIXEL_RATIO = 0.01
BOX_MIN_W         = 60
BOX_MAX_H         = 80
BOX_MIN_ASPECT    = 1.5

# ─────────────────────────────────────────────
# OPENCV PIPELINE  (Advanced variant)
# ─────────────────────────────────────────────

def load_gray(path):
    return np.array(Image.open(path).convert("L"))

def preprocess(gray):
    _, binary = cv2.threshold(gray, 0, 255,
                              cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    return cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

def opencv_checkboxes(binary, img_w, img_h):
    contours, _ = cv2.findContours(binary, cv2.RETR_CCOMP,
                                   cv2.CHAIN_APPROX_SIMPLE)
    boxes = []
    for cnt in contours:
        peri   = cv2.arcLength(cnt, True)
        approx = cv2.approxPolyDP(cnt, 0.02 * peri, True)
        if len(approx) != 4:
            continue
        x, y, w, h = cv2.boundingRect(approx)
        if w > img_w * 0.95 or h > img_h * 0.95:
            continue
        sq = abs(1 - w / h) if h else 99
        if CHECKBOX_MIN <= w <= CHECKBOX_MAX and \
           CHECKBOX_MIN <= h <= CHECKBOX_MAX and sq <= SQUARE_TOLERANCE:
            boxes.append((x, y, x + w, y + h))          # xyxy
    return boxes

def opencv_lines(binary, checkboxes_xyxy):
    kernel  = cv2.getStructuringElement(cv2.MORPH_RECT, (40, 1))
    h_lines = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
    contours, _ = cv2.findContours(h_lines, cv2.RETR_EXTERNAL,
                                   cv2.CHAIN_APPROX_SIMPLE)
    lines = []
    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        if w < LINE_MIN_W or h > LINE_MAX_H:
            continue
        lx1 = max(0, x - LABEL_SEARCH_PX)
        lx2 = max(0, x - 5)
        ly1 = max(0, y - 10)
        ly2 = min(binary.shape[0], y + h + 10)
        if lx2 <= lx1:
            continue
        region = binary[ly1:ly2, lx1:lx2]
        if region.size == 0:
            continue
        if np.count_nonzero(region) / region.size <= LABEL_PIXEL_RATIO:
            continue
        VICINITY = 20
        near_cb = any(
            x >= cx - VICINITY and x + w <= cx + (ex - cx) + VICINITY and
            y >= cy - VICINITY and y + h <= cy + (ey - cy) + VICINITY
            for (cx, cy, ex, ey) in checkboxes_xyxy
        )
        if near_cb:
            continue
        lines.append((x, y, x + w, y + h))
    return lines

def opencv_boxes(binary, img_w, img_h):
    contours, _ = cv2.findContours(binary, cv2.RETR_CCOMP,
                                   cv2.CHAIN_APPROX_SIMPLE)
    boxes = []
    for cnt in contours:
        peri   = cv2.arcLength(cnt, True)
        approx = cv2.approxPolyDP(cnt, 0.02 * peri, True)
        if len(approx) != 4:
            continue
        x, y, w, h = cv2.boundingRect(approx)
        if w > img_w * 0.95 or h > img_h * 0.95:
            continue
        aspect = w / h if h else 0
        if BOX_MIN_W <= w and h <= BOX_MAX_H and aspect >= BOX_MIN_ASPECT:
            boxes.append((x, y, x + w, y + h))
    return boxes

def run_opencv(img_path):
    """Return {0: [xyxy,...], 1: [...], 2: [...]} from OpenCV Advanced."""
    gray   = load_gray(img_path)
    h, w   = gray.shape
    binary = preprocess(gray)

    cbs   = opencv_checkboxes(binary, w, h)
    lines = opencv_lines(binary, cbs)
    bxs   = opencv_boxes(binary, w, h)

    return {0: cbs, 1: lines, 2: bxs}

# ─────────────────────────────────────────────
# GEOMETRY HELPERS
# ─────────────────────────────────────────────

def iou(a, b):
    """Intersection-over-Union of two xyxy boxes."""
    ix1 = max(a[0], b[0]); iy1 = max(a[1], b[1])
    ix2 = min(a[2], b[2]); iy2 = min(a[3], b[3])
    inter = max(0, ix2 - ix1) * max(0, iy2 - iy1)
    if inter == 0:
        return 0.0
    area_a = (a[2]-a[0]) * (a[3]-a[1])
    area_b = (b[2]-b[0]) * (b[3]-b[1])
    return inter / (area_a + area_b - inter)

def best_iou(box, candidates):
    """Return the highest IoU between box and any candidate."""
    if not candidates:
        return 0.0
    return max(iou(box, c) for c in candidates)

def geometric_filter(box, cls_id, img_w, img_h):
    """
    Last-resort geometric check for unresolved low-conf YOLO boxes.
    Mirrors the same rules used in OpenCV classification.
    """
    x1, y1, x2, y2 = box
    w = x2 - x1; h = y2 - y1
    if w <= 0 or h <= 0:
        return False
    if w > img_w * 0.95 or h > img_h * 0.95:
        return False

    if cls_id == 0:   # checkbox
        sq = abs(1 - w / h)
        return (CHECKBOX_MIN <= w <= CHECKBOX_MAX and
                CHECKBOX_MIN <= h <= CHECKBOX_MAX and
                sq <= SQUARE_TOLERANCE)
    elif cls_id == 1: # line
        aspect = w / h if h else 0
        return w >= LINE_MIN_W and h <= LINE_MAX_H * 3   # slight leniency
    elif cls_id == 2: # box
        aspect = w / h if h else 0
        return w >= BOX_MIN_W and h <= BOX_MAX_H and aspect >= BOX_MIN_ASPECT
    return False

# ─────────────────────────────────────────────
# HYBRID FUSION
# ─────────────────────────────────────────────

def hybrid_detect(img_path, model):
    """
    Fuse YOLOv11 and OpenCV Advanced detections.

    Strategy:
      1. YOLO conf >= CONF_HIGH          → accept directly.
      2. CONF_LOW <= conf < CONF_HIGH    → cross-ref with OpenCV;
             IoU >= IOU_CONFIRM          → accept
             else geometric_filter pass  → accept
             else                        → discard
      3. conf < CONF_LOW                 → discard.
      4. OpenCV candidates with no YOLO overlap (any conf) → supplement.
    """
    img_w, img_h = Image.open(img_path).size

    # ── Step A: YOLO inference ────────────────
    results = model.predict(img_path, conf=CONF_LOW, verbose=False)[0]

    high_conf  = {c: [] for c in CLASS_NAMES}   # accepted outright
    low_conf   = {c: [] for c in CLASS_NAMES}   # need verification
    all_yolo   = {c: [] for c in CLASS_NAMES}   # every yolo box (for overlap check)

    for box in results.boxes:
        cls_id = int(box.cls)
        conf   = float(box.conf)
        coords = tuple(float(v) for v in box.xyxy[0])
        all_yolo[cls_id].append(coords)
        if conf >= CONF_HIGH:
            high_conf[cls_id].append(coords)
        else:
            low_conf[cls_id].append(coords)

    # ── Step B: OpenCV candidates ─────────────
    opencv_preds = run_opencv(img_path)

    final = {c: list(high_conf[c]) for c in CLASS_NAMES}

    # ── Step C: verify low-conf YOLO boxes ────
    for cls_id in CLASS_NAMES:
        for box in low_conf[cls_id]:
            ov_iou = best_iou(box, opencv_preds[cls_id])
            if ov_iou >= IOU_CONFIRM:
                # OpenCV confirms the detection
                final[cls_id].append(box)
            elif geometric_filter(box, cls_id, img_w, img_h):
                # Passes shape-based rules → keep
                final[cls_id].append(box)
            # else → discard

    # ── Step D: supplement with OpenCV-only detections ────
    for cls_id in CLASS_NAMES:
        for cv_box in opencv_preds[cls_id]:
            # Only add if not already covered by any YOLO detection
            if best_iou(cv_box, all_yolo[cls_id]) < IOU_CONFIRM:
                final[cls_id].append(cv_box)

    return final

# ─────────────────────────────────────────────
# EVALUATION  (same tolerance-based logic)
# ─────────────────────────────────────────────

def load_gt(path, img_w, img_h):
    data = json.load(open(path))
    gt   = {0: [], 1: [], 2: []}
    for cx, cy in data.get("checkboxes", []):
        s = 0.01 * min(img_w, img_h)
        gt[0].append((cx-s, cy-s, cx+s, cy+s))
    for x1,y1,x2,y2 in data.get("lines", []):
        px, py = 0.01*img_w, 0.01*img_h
        gt[1].append((min(x1,x2)-px, min(y1,y2)-py,
                      max(x1,x2)+px, max(y1,y2)+py))
    for x1,y1,x2,y2 in data.get("boxes", []):
        gt[2].append((x1,y1,x2,y2))
    return gt

def match_boxes(gt_boxes, pred_boxes, tol_x, tol_y):
    candidates = []
    for gi, (gx1,gy1,gx2,gy2) in enumerate(gt_boxes):
        for pi, (px1,py1,px2,py2) in enumerate(pred_boxes):
            if (abs(px1-gx1)<=tol_x and abs(py1-gy1)<=tol_y and
                abs(px2-gx2)<=tol_x and abs(py2-gy2)<=tol_y):
                dist = math.sqrt((gx1-px1)**2+(gy1-py1)**2) + \
                       math.sqrt((gx2-px2)**2+(gy2-py2)**2)
                candidates.append((dist, gi, pi))
    candidates.sort(key=lambda x: x[0])
    matched_gt, matched_pred, count = set(), set(), 0
    for _, gi, pi in candidates:
        if gi not in matched_gt and pi not in matched_pred:
            matched_gt.add(gi); matched_pred.add(pi); count += 1
    return count

def compute_metrics(matches, total_pred, total_gt):
    p   = matches / total_pred if total_pred else 0
    r   = matches / total_gt   if total_gt   else 0
    f1  = (2*p*r / (p+r))     if (p+r)      else 0
    den = total_gt + total_pred - matches
    acc = matches / den        if den        else 0
    return p, r, f1, acc

# ─────────────────────────────────────────────
# MAIN
# ─────────────────────────────────────────────

def main():
    model = YOLO(WEIGHTS_PATH)

    gt_stems  = {f[:-5] for f in os.listdir(GT_FOLDER)    if f.endswith(".json")}
    img_stems = {f[:-4]  for f in os.listdir(IMAGE_FOLDER) if f.endswith(".png")}
    stems     = sorted(gt_stems & img_stems)

    if not stems:
        print("No matching GT + image pairs found. Check your paths.")
        return

    print(f"Evaluating hybrid framework on {len(stems)} documents …\n")

    for tol in TOLERANCES:
        totals = {cls: [0, 0, 0] for cls in CLASS_NAMES}  # [gt, pred, match]

        print("=" * 60)
        print(f"TOLERANCE: {int(tol*100)}%")
        print("=" * 60)

        for stem in stems:
            img_path = os.path.join(IMAGE_FOLDER, stem + ".png")
            gt_path  = os.path.join(GT_FOLDER,    stem + ".json")
            img_w, img_h = Image.open(img_path).size
            tol_x, tol_y = img_w * tol, img_h * tol

            gt   = load_gt(gt_path, img_w, img_h)
            pred = hybrid_detect(img_path, model)

            # ── Save fused predictions as JSON ────────
            out = {
                "checkboxes": [[x1,y1,x2,y2] for x1,y1,x2,y2 in pred[0]],
                "lines":      [[x1,y1,x2,y2] for x1,y1,x2,y2 in pred[1]],
                "boxes":      [[x1,y1,x2,y2] for x1,y1,x2,y2 in pred[2]],
            }
            with open(os.path.join(OUTPUT_FOLDER, stem + ".json"), "w") as f:
                json.dump(out, f, indent=2)

            for cls_id in CLASS_NAMES:
                matches = match_boxes(gt[cls_id], pred[cls_id], tol_x, tol_y)
                totals[cls_id][0] += len(gt[cls_id])
                totals[cls_id][1] += len(pred[cls_id])
                totals[cls_id][2] += matches

        for cls_id, name in CLASS_NAMES.items():
            gt_n, pred_n, match_n = totals[cls_id]
            p, r, f1, acc = compute_metrics(match_n, pred_n, gt_n)
            print(f"\n{name}:")
            print(f"  Precision: {p:.3f} | Recall: {r:.3f} | "
                  f"F1: {f1:.3f} | Accuracy (Jaccard): {acc:.3f}")

        print()


if __name__ == "__main__":
    main()
```
