```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the paper (Tolerance: 20%)
models = ['OpenCV', 'YOLOv8', 'YOLOv11', 'YOLOv26-s', 'YOLOv26-l']

# F1 Scores for each class
checkbox_f1 = [0.699, 0.823, 0.827, 0.750, 0.725]
lines_f1 = [0.454, 0.808, 0.836, 0.785, 0.821]
boxes_f1 = [0.520, 0.724, 0.760, 0.724, 0.767]

x = np.arange(len(models))
width = 0.25

fig, ax = plt.subplots(figsize=(14, 7))

# Plotting the bars
rects1 = ax.bar(x - width, checkbox_f1, width, label='Checkboxes', color='#3182bd')
rects2 = ax.bar(x, lines_f1, width, label='Lines', color='#1a9850')
rects3 = ax.bar(x + width, boxes_f1, width, label='Boxes', color='#b35806')

# Adding labels on top of the bars with bold font
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.3f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9, fontweight='bold')

autolabel(rects1)
autolabel(rects2)
autolabel(rects3)

# Formatting - All fonts set to bold
ax.set_ylabel('F1 score', fontsize=12, fontweight='bold')
ax.set_title('F1 Score by Model and Element Class (20% Tolerance)', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(models, fontsize=11, fontweight='bold')

# Bold the y-axis tick labels
plt.setp(ax.get_yticklabels(), fontweight='bold')

ax.set_ylim(0, 1.0)

# Legend with bold font
ax.legend(prop={'weight': 'bold'})

# Style adjustments to match original
ax.yaxis.grid(True, linestyle='-', which='major', color='lightgrey', alpha=0.5)
ax.set_axisbelow(True)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('lightgrey')
ax.spines['bottom'].set_color('lightgrey')

plt.tight_layout()
plt.savefig('f1_score_plot_bold.png')
print("Plot saved as f1_score_plot_bold.png")
```
